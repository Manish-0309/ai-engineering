Operators are symbols that perform some operation on one or more operands.

```python
# Arithmetic Operators : Used for mathematical operations like addition, subtraction, multiplication, etc.
num1 = 5
num2 = 3
print("Addition :" , num1 + num2)
print("Subtraction :" , num1 - num2)
print("Multiplication :" , num1 * num2)
print("Division :" , num1 / num2)
print("Floor Division :" , num1 // num2)
print("Remainder(mod) :" , num1 % num2)
print("Exponent :" , num1 ** num2)
```

```python
# Assignment Operators : Used for assigning values
x = 5
x = x + 5 
x += 5  #shorthand

# same can be done for all other arithmetic operators

print(x)
```

```python
# Comparison Operators: Used to compare values
x = 8
y = 6
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
# Logical Operators : Used for boolean logic
# and : return True if all the conditions are true
a = True
b = False

print(a and b)

# or: return True if atleast one condition is true
print(a or b)

# not: reverse the output
print(not b)
print(not a)
```

```python
# short circuiting 'and'
a, b = 5, 1
print(a and b) #return real number for real number
```

```python
# Bitwise Operators: Operates on binary representation
# AND - &
print(5 & 1)

# OR - |
print(5 | 3)

# NOT - ~
print(~1)

# XOR - ^
print(5 ^ 7)

# left shift
print(5 << 1)

# right shift
print(5 >> 1)
```

```python
# Membership Operators : return True if specified object is present in the sequence or collection
# in : return True if object is present 
# not in: return True if object is not present in the sequence

# works with : list, tuple, set, string, dictionary(check only keys)

fruits = ['Apple', 'Banana', 'Orange', 'Grapes']
print(fruits)

print('Apple' in fruits)
print('Avacado' in fruits)
```

```python
# Identity Operators: check if two vars/object have same memory reference
# is: return True if they have same memory
# is not: return True if they do not have same memory

a = 5
b = a
c = 8

print(a is b)
print(a is not c)
```
