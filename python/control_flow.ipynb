```python
# Conditionals : decision making
# if, elif, else

if 5 > 6:
    print("Hello")
elif 5 < 8:
    print("Hi")
else:
    print("Bye")

# else : fallback
```

```python
# Truthiness: Object is considered Truthy if it evaluates to 'true' and 'falsy' if it evaluates to 'false'

# Truthy values: non-zero numbers, non-empty string, non-empty list,tuple,dict, set
# Falsy values: None, False, Empty sequence, collections 
```

```python
# Loop: Loop is used to perform a task repeatedly until a particular condition is satisfied. 
# Python mainly provides: 
        # for loop     # while loop

# for: iterates over iterable

for i in range(5):
    print(i) 

for j in range(2, 6):
    print(j, end="")

for k in range(10, 0, -1):
    print(k, end=" ")
```

```python
# while loop: runs while condition is true.
i = 0
while i < 5:
    print("Manish", end=" ")
    i += 1
```

```python
# controlling execution
# break: terminate loop immediately
for i in range(5):
    if i == 5:
        break
    print(i)
```

```python
# continue: skip iteration
for i in range(3):
    if i == 0:
        continue
    print(i)
```

```python
# pass: placeholder statement (mostly used in development)
for i in range(5):
    pass
```
