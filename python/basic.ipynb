# Python

Python is a high-level, interpreted, general-purpose programming language.

## Key Characteristics
- **High-level**: Automatic memory management
- **Interpreted**: Code is executed line by line (CPython - default, PyPy, Jython)
- **General-purpose**: Backend, scripting, AI/ML, automation, and more
- **Multi-Paradigm**: Supports OOP, Functional, and Procedural programming
- **Dynamically typed**: Variables can hold any data type

## History
- **Developed by**: Guido Van Rossum
- **Year**: 1991

```python
# variables
a = 5
print(a) #In python variables are dynamically typed, no need to ecplicitly define the datatype.

# multiple assignments
x, y = 5, 6 # tuple unpacking
x=y=z = 9  # same value to different variables
x, y = y, x # swapped values
print(x, y, z)
```

```python
del x #delete variable
```

```python
# Datatype : In Python type is associated with object
    # Built in Datatypes:
# Numeric Type : int, float, complex
# Boolean Type: True, False
# Sequence Type: list(mutable), tuple, range
# Mapping Type: dictionary (mutable)
# Set Type: set (mutable)
# None Type: None : represents absence of value 
```

```python
n = 5
print(type(n)) # type() will print the datatype of variables.
print(type(3.14))
print(type(3 + 4j))
print(type("Manish"))
print(type([1, 2, 3]))
print(type((4, 5, 6)))
print(type(range(5)))
print(type({}))
print(type(True))
print(type(None))
```

```python
# User input
name = input("Enter Your Name: ")
print(f"{name}")

# By default it will treat every input as str, we have to typecast them.
# Implicit: Smaller into larger datatype -- no data loss
# Explicit: Larger into smaller datatype - data loss may occur

x = 2 + 9.0
print(x)

f = int(5.0)
print(f)

# User input with typecasting.
age = int(input("Enter Your age: "))
print(age)
```
